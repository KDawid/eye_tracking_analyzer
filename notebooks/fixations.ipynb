```python
from eye_tracking.fuse_eyes import fuse_raw_data, EyeFusionProtocol
from pathlib import Path
import polars as pl
from eye_tracking.columns import Column
from eye_tracking.reader import read_raw_data, read_demo_data
from eye_tracking.input_formats import SIMPLEGAZETRACKER, TOBII

tobii_raw = read_raw_data(Path("data/1_rawdata.csv"), fmt=TOBII)
tobii_df = fuse_raw_data(tobii_raw, EyeFusionProtocol.Any)

sgt_raw = read_raw_data(Path("../notebooks/simplegazetracker_dummy.csv"), fmt=SIMPLEGAZETRACKER)
sgt_df = fuse_raw_data(sgt_raw, EyeFusionProtocol.Any)

demo_df = read_demo_data("../notebooks/data/demo_data.txt")
```

```python
from eye_tracking.fixation.eyefeatures import idt

for data_source, source_df in [("Tobii", tobii_df), ("SGT", sgt_df), ("Demo", demo_df)]:
    fixations = idt(source_df, min_duration=0.1, max_duration=1.0, max_dispersion=0.05)
    print(f"{data_source} - Fixations:")
    print(fixations)
    print()
```

```python
from eye_tracking.fixation.eyefeatures import ivt

for data_source, source_df in [("Tobii", tobii_df), ("SGT", sgt_df), ("Demo", demo_df)]:
    fixations = ivt(source_df, velocity_threshold=0.5, min_duration=0.1)
    print(f"{data_source} - Fixations:")
    print(fixations)
    print()
```

```python
from eye_tracking.fixation.eyefeatures import ihmm

for data_source, source_df in [("Tobii", tobii_df), ("SGT", sgt_df), ("Demo", demo_df)]:
    fixations = ihmm(source_df)
    print(f"{data_source} - Fixations:")
    print(fixations)
    print()
```
